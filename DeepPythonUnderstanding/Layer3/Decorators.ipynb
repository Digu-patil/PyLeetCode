# Core components of a decorator
1. Functions are objects
2. Functions can be passed as arguments
3. Function can be returned from another function

This is the core idea of a decorator

```python
def outer(func):
    return func

def inner():
    print("You executed inner function")

f = outer(inner)
print(f)
f()
```

## Decorator is a function that wraps another function

```python
# Manually writing a decorator

def decorator(func):
    def outer():
        print('This is outer function')
        func()
        print('This is post func() execution')
    return outer

def f():
    print('f is executed')

x = decorator(f)
print(x())
print(x.__closure__)
```

```python
'''
Instead of writing this
x = decorator(new_func)
print(x())
We define decorators
'''

@decorator
def newfunc():
    print('New Func is executed')

newfunc()
```

```python
def fetch_decorator(func):
    def logging(*args, **kwargs):
        print(f"Calling {func.__name__}")
        result1 = func(*args, **kwargs)
        print(result1)
        print(f"Finished {func.__name__}")
        return result1
    return logging

@fetch_decorator
def fetch_user():
    return {"id":1}
    # return result

@fetch_decorator
def fetch_orders():
    return [1,2,3]
    # return result

fetch_user()
fetch_orders()
```

```python
# Lets say we wanted to time function execution
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        f = func(*args, **kwargs)
        end = time.time()
        print(f'{func.__name__} took {int(end-start)}s to execute')
        return func
    return wrapper

@timer
def func1():
    time.sleep(5)

@timer
def func2():
    time.sleep(10)

func1()
func2()
```
