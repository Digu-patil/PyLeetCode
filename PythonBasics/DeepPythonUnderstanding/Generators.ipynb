# Generators

How can I produce values one by one, on demand without computing everything upfront?
- This is what generators are for
- instead of return we use yield keyword to create "Generators"

```python
# Here is an example of a simple generator function
# This is a counter generator which will count till the stop_val.

def counter(stop_val):
    count = 0
    while count < stop_val:
        yield count
        count+=1

inc = counter(3) # this creates a generator object
print(next(inc)) # 0
print(next(inc)) # 1
print(next(inc)) # 2
print(next(inc)) # StopIteration Exeception is raised
```

## Underlying Workings of a Generator 

Let's go step by step to understand how thing woked in the above example
1. A function object "counter" was created.
2. This function object lives on the heap
3. Please note that "counter" is a function object and "coutner()" i.e., calling the function object created a generator object
    1. Python sees "counter" contains yield
    2. And it returns generator body.
    3. Important **calling a funciton does not execute it
4. inc = counter(3) doing this we are only creating a generator object and "inc" is referencing to that generator object.
5. next(inc) - Calling "next" creates a generator frame and starts execution
6. Local variable counter is created in the generator frame.
7. When "yield" executes, Python suspends/pauses the frame instead of destorying it.
8. On the next "next(inc)", execution resumens from where it has stopped.
9. Because the same generator frame is being reused, counter keeps on increasing.

```python
print(type(counter)) # This will return a function object
print(type(counter(3))) # This will return a generator object, note that execution has not yet started
```


Important Point to note
1. Everytime we use "counter()" it creates a new generator frame
2. so next(counter()) will give you the same default value everytime, as new generator frame is being created everytime.
3. x = counter() and then next(x) works
4. This works because x is pointing to the generator object and when we do next(x) is executes data in the same frame and does not create a new frame.
