# What are Iterators?

> Iterator is an object that can give you the next value on demand.

Python has a __next__() method for this, once this is exhausted it raises StopIteration Error.

## Difference between Iterator and Iterable.

Iterable is an object on which we can create an iterator.
i.e 
1. We have a list [1,2,3], this would be of <class list>
2. We can convert this into a iterator object using "iter"
3. Hence this list becomes an iterable, and the object created using "iter" is called the iterator.
4. Once the iterator is created, we can access the next element using "next" function.

**IMPORTANT
Anything that supports a for loop, is an iterable and we can create a iterator object out of it.

```python
list = [1,2,3]
print(type(list))

iter_list = iter(list)
print(type(iter_list)) # this is an iterator object
```

```python
# Accessing the elements of the list_iterator

print(next(iter_list))
print(next(iter_list))
```

# How for loops are internally implemented in Python

A python for loop is essentially a while True loop that repeatedly calls next() until it exits using the StopIeration.

```python
# Implementing a loop equivalent of and iterator

# prints all the elements of the list

name = "Digu"
iter_name = iter(name)
print("All elements of the list")
for i in name:
    print(i)
print("----")

'''
Now we know next element of an iterator object can be accessed using "next" function
This is exactly how a for loop is executed internally in python
'''
print("How for loop works internally")
while True:
    try:
        x = next(iter_name) # iter_list is an iterator object
    except StopIteration:
        break
    print(x)


```

# Generators V/S Iterators

- Generators are a special implementation of iterators.
- Consider Generator to be a shortcut implementation of iterators.
- Generaotrs are compiler-generated iterators.
- With Generators we do no manullay define

|Generator|Iterator|
|---------|--------|
|Python Manages state|We manually manage state|
|Clean, readable | We write the boiler plate code|
|Safer|Error Prone|

```python
# Let's use our "counter" generator for this

def counter(stop_val):
    ct = 0
    while ct <= stop_val: 
        yield ct
        ct += 1

ct = counter(2)
print(type(ct))

# Let's try to see if the generator is a iterator as well
print(iter(ct) is ct)
# This prints true, stating that an iterator is a generator as well
```
