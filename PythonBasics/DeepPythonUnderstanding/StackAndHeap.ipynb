```python
# Inspceting where the variables or namespaces live

a = 20
b = [20,60,80]

def outer(a):
    x = 10
    outer_frame = inspect.currentframe()
    print(f'These are local var to outer function {outer_frame.f_locals}')
    print(f'These are global var to outer function {outer_frame.f_globals}')
    print(f'Stack frame created for the following function {outer_frame.f_locals.co_name}')
    def inner():
        p = a + x
        inner_frame = inspect.currentframe()
        print(f'These are local var to outer function {inner_frame.f_locals}')
        print(f'These are global var to outer function {inner_frame.f_globals}')
        return p
    return inner

b = outer(a)
```
